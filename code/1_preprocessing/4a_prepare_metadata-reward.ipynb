```python
DATA_DIR = "" # Write your data path here
WORKING_DIR = "" # Write your working path here
FD_THRESH = 0.5
ALLOW_NA_FD = True

DSET_IMG_MAP = {
    "antGainVsantControl": {
        "integramoods": op.join(
            DATA_DIR,
            "integramoods/halfpipe/*/*/derivatives/halfpipe/{subject}/func/task-reward/{subject}_task-reward_feature-rewardVar1_taskcontrast-antGainVsAntControl_stat-z_statmap.nii.gz",
        ),
        "dynamore": op.join(
            DATA_DIR,
            "/mnt/datasets/stressimaging/dynamore/halfpipe/mid/derivatives/halfpipe/{subject}/func/task-mid/{subject}_task-mid_feature-midAnticipationAroma_taskcontrast-antWinGtAntZero_stat-z_statmap.nii.gz",
        ),
        "abcd": op.join(
            DATA_DIR,
            "abcd/halfpipe/*/derivatives/halfpipe-fixed-effects/{subject}/*/func/task-mid/{subject}_*_task-mid_feature-mid1_taskcontrast-antGainVsAntControl_stat-z_statmap.nii.gz",
        ),
        "abcd-2025": op.join(
            DATA_DIR,
            "abcd/halfpipe-2025/derivatives/halfpipe-fixed-effects/{subject}/ses-{fu}/func/task-mid/{subject}_ses-{fu}_task-mid_feature-mid1_taskcontrast-antGainVsAntControl_stat-z_statmap.nii.gz",
        ),
        "imagen": op.join(
            DATA_DIR,
            "imagen/halfpipe/{fu}/*/derivatives/halfpipe/{subject}/*/func/task-mid/{subject}_*_task-mid_feature-midVar1_taskcontrast-antGainVsAntControl_stat-z_statmap.nii.gz",
        ),
    },
    "outHitVsMiss": {
        "integramoods": op.join(
            DATA_DIR,
            "integramoods/halfpipe/*/*/derivatives/halfpipe/{subject}/func/task-reward/{subject}_task-reward_feature-rewardVar1_taskcontrast-outHitVsMiss_stat-z_statmap.nii.gz",
        ),
        "hcpya": op.join(
            DATA_DIR,
            "hcp/halfpipe/young_adult/derivatives/halfpipe/{subject}/ses/func/task-GAMBLING/{subject}_task-GAMBLING_feature-GAMBLINGVar1_taskcontrast-outHitVsMiss_stat-z_statmap.nii.gz",
        ),
        "hcpd": op.join(
            DATA_DIR,
            "hcp/halfpipe/development/derivatives/halfpipe/{subject}/ses/func/task-GUESSING/{subject}_task-GUESSING_feature-guessingVar1_taskcontrast-outGainVsOutLoss_stat-z_statmap.nii.gz",
        ),
        "chcp": op.join(
            DATA_DIR,
            "hcp/raw/chinese/preproc/{subject}/MNINonLinear/Results/tfMRI_Gambling/tfMRI_Gambling_hp200_s2_level2.feat/StandardVolumeStats/cope6.feat/zstat1_s6.nii.gz",
        ),
        "dynamore": op.join(
            DATA_DIR,
            "stressimaging/dynamore/halfpipe/mid/derivatives/halfpipe/{subject}/func/task-mid/{subject}_task-mid_feature-midFeedBackAroma_taskcontrast-hitGtMiss_stat-z_statmap.nii.gz",
        ),
        "abcd": op.join(
            DATA_DIR,
            "abcd/halfpipe/*/derivatives/halfpipe-fixed-effects/{subject}/*/func/task-mid/{subject}_*_task-mid_feature-mid1_taskcontrast-outHitVsMiss_stat-z_statmap.nii.gz",
        ),
        "abcd-2025": op.join(
            DATA_DIR,
            "abcd/halfpipe-2025/derivatives/halfpipe-fixed-effects/{subject}/ses-{fu}/func/task-mid/{subject}_ses-{fu}_task-mid_feature-mid1_taskcontrast-outHitVsMiss_stat-z_statmap.nii.gz",
        ),
        "ds003858": op.join(
            DATA_DIR,
            "openneuro/ds003858/halfpipe/derivatives/halfpipe/{subject}/func/task-MID/{subject}_task-MID_run-8_feature-midVar1_taskcontrast-hitVsMiss_stat-z_statmap.nii.gz",
        ),
        "ds005479": op.join(
            DATA_DIR,
            "openneuro/ds005479/halfpipe/derivatives/halfpipe/{subject}/func/task-MID/{subject}_task-MID_feature-midVar1_taskcontrast-outGainVsOutLoss_stat-z_statmap.nii.gz",
        ),
    },
    "outGainVsOutNoGain": {
        "integramoods": op.join(
            DATA_DIR,
            "integramoods/halfpipe/*/*/derivatives/halfpipe/{subject}/func/task-reward/{subject}_task-reward_feature-rewardVar1_taskcontrast-outGainVsOutNoGain_stat-z_statmap.nii.gz",
        ),
        "dynamore": op.join(
            DATA_DIR,
            "/mnt/datasets/stressimaging/dynamore/halfpipe/mid/derivatives/halfpipe/{subject}/func/task-mid/{subject}_task-mid_feature-midFeedBackAroma_taskcontrast-gainGtNoGain_stat-z_statmap.nii.gz",
        ),
        "abcd": op.join(
            DATA_DIR,
            "abcd/halfpipe/*/derivatives/halfpipe-fixed-effects/{subject}/*/func/task-mid/{subject}_*_task-mid_feature-mid1_taskcontrast-outGainVsOutNoGain_stat-z_statmap.nii.gz",
        ),
        "abcd-2025": op.join(
            DATA_DIR,
            "abcd/halfpipe-2025/derivatives/halfpipe-fixed-effects/{subject}/ses-{fu}/func/task-mid/{subject}_ses-{fu}_task-mid_feature-mid1_taskcontrast-outGainVsOutNoGain_stat-z_statmap.nii.gz",
        ),
        "ds005479": op.join(
            DATA_DIR,
            "openneuro/ds005479/halfpipe/derivatives/halfpipe/{subject}/func/task-MID/{subject}_task-MID_feature-midVar1_taskcontrast-outGainVsOutNoGain_stat-z_statmap.nii.gz",
        ),
        "ds003858": op.join(
            DATA_DIR,
            "openneuro/ds003858/halfpipe/derivatives/halfpipe/{subject}/func/task-MID/{subject}_task-MID_run-8_feature-midVar1_taskcontrast-outGainVsOutNoGain_stat-z_statmap.nii.gz",
        ),
        "imagen": op.join(
            DATA_DIR,
            "imagen/halfpipe/{fu}/*/derivatives/halfpipe/{subject}/*/func/task-mid/{subject}_*_task-mid_feature-midVar1_taskcontrast-outGainVsOutNoGain_stat-z_statmap.nii.gz",
        ),
    },
}


def filter_task(df, task_name, add_na=False, fd_thresh=FD_THRESH):
    """
    Filter rows for a given task with good rating and FD threshold.

    Parameters
    ----------
    df : pandas.DataFrame
        Input dataframe.
    task_name : str
        Task name to filter on.
    add_na : bool, optional
        If True, allow rows with NA `fd_mean` in addition to rows below threshold.

    Returns
    -------
    pandas.DataFrame
        Filtered dataframe.
    """
    mask = (df["task"] == task_name) & (df["rating"] == "good")
    if add_na:
        mask &= (df["fd_mean"] < fd_thresh) | (df["fd_mean"].isna())
    else:
        mask &= df["fd_mean"] < fd_thresh
    return df.loc[mask]


def aggregate_qc(
    qc_df: pd.DataFrame,
    by: Sequence[str],
) -> pd.DataFrame:
    """
    Aggregate QC metrics by group.

    Parameters
    ----------
    qc_df : pandas.DataFrame
        Input QC dataframe with columns `rating_col` and `fd_col`.

    Returns
    -------
    pandas.DataFrame
        Aggregated dataframe with:
        - `fd_mean`: mean of `fd_col`
        - `rating`: "good" if all ratings in group are "good" (case/space-insensitive), else "bad"
    """

    def _agg_rating(s: pd.Series) -> str:
        s = s.dropna().astype(str).str.strip().str.lower()
        return "good" if (len(s) > 0 and (s == "good").all()) else "bad"

    return (
        qc_df.groupby(list(by), dropna=False)
        .agg(
            fd_mean=("fd_mean", "mean"),
            rating=("rating", _agg_rating),
        )
        .reset_index()
    )


def process_dataset_contrasts(
    metadata_dset: pd.DataFrame,
    dset: str,
    contrasts: Sequence[str],
    dset_img_map: Dict[str, Dict[str, str]],
    metadata_by_contrast: Dict[str, Dict[str, pd.DataFrame]],
    fpath_col: str = "fpath",
    dropna_subset: Sequence[str] | None = ("fpath",),
    print_counts: bool = True,
) -> None:
    """
    Attach per-subject files for each contrast and store results in `metadata_by_contrast`.

    Parameters
    ----------
    metadata_dset : pandas.DataFrame
        Preprocessed metadata for one dataset; must contain columns required by path templates
        (e.g., `subject`, optionally `fu`).
    dset : str
        Dataset key (e.g., "integramoods", "abcd").
    contrasts : Sequence[str]
        Contrast names to process (keys of `dset_img_map`).
    dset_img_map : Dict[str, Dict[str, str]]
        Mapping from contrast -> dataset -> glob pattern template containing placeholders
        like `{subject}` and optionally `{fu}`.
    metadata_by_contrast : Dict[str, Dict[str, pandas.DataFrame]]
        Output accumulator. This function assigns: metadata_by_contrast[contrast][dset] = df.
    fpath_col : str
        Column name to store matched file paths.
    dropna_subset : Sequence[str] | None
        Columns used to drop rows with missing values after file matching. Use None to skip drop.
    print_counts : bool
        If True, prints the number of rows per contrast after filtering missing files.

    Returns
    -------
    None
    """
    for cont in contrasts:
        df = metadata_dset.copy()
        tmpl = dset_img_map[cont][dset]
        keys = re.findall(r"{(\w+)}", tmpl)

        files: list[str | None] = []
        for _, row in df.iterrows():
            try:
                path = tmpl
                for k in keys:
                    path = path.replace("{" + k + "}", str(row[k]))
                files.append(glob(path)[0])
            except Exception:
                files.append(None)

        df[fpath_col] = files
        if dropna_subset:
            df = df.dropna(subset=list(dropna_subset))

        metadata_by_contrast[cont][dset] = df

        if print_counts:
            print(f"Number of subjects in {cont} for {dset}: {len(df)}")
```

```python
def hist_by(df: pd.DataFrame, x: str, hue: str, title: str, xlabel: str) -> None:
    """
    Histogram with stacked hue.

    Parameters
    ----------
    df : pandas.DataFrame
    x : str
        Column to plot on x-axis.
    hue : str
        Column for hue.
    title : str
    xlabel : str
    """
    _, ax = plt.subplots(figsize=(15, 10), dpi=300)

    # determine categories in the same order they appear
    sites = df[hue].dropna().astype(str).unique().tolist()

    # use a qualitative palette designed for many distinct categories
    # 'husl' spaces hues evenly and remains fairly distinguishable up to many colors
    palette = sns.color_palette("husl", n_colors=len(sites))

    sns.histplot(
        data=df,
        x=x,
        hue=hue,
        bins=60,
        ax=ax,
        stat="count",
        multiple="stack",
        common_norm=False,
        alpha=0.85,
        linewidth=1.0,
        edgecolor="black",
        palette=palette,
    )

    handles, labels = ax.get_legend_handles_labels()

    # If seaborn didn't produce a usable legend, build one from our palette
    if not labels or len(labels) != len(sites):
        handles = [
            Patch(facecolor=palette[i], edgecolor="black", alpha=0.85)
            for i in range(len(sites))
        ]
        labels = sites

    ax.legend(
        handles,
        labels,
        loc="upper center",
        bbox_to_anchor=(0.5, -0.2),
        ncol=5,
        frameon=False,
    )
    plt.subplots_adjust(bottom=0.18)
    ax.set_xlabel(xlabel, fontsize=16, labelpad=12)
    ax.set_ylabel("Count", fontsize=16, labelpad=12)
    ax.set_title(title, fontsize=20, pad=20)
    ax.tick_params(axis="both", which="major", labelsize=13)
    ax.xaxis.set_major_locator(mticker.MaxNLocator(nbins=25))
    ax.yaxis.set_major_locator(mticker.MaxNLocator(nbins=6))
    sns.despine(left=True, bottom=True)
    plt.tight_layout()
    plt.show()


def plot_var_grid(df: pd.DataFrame, vars_to_plot: List[str], ncols: int = 3) -> None:
    """
    Small-multiples bar counts for many variables.

    Parameters
    ----------
    df : pandas.DataFrame
    vars_to_plot : list of str
    ncols : int
    """
    n_vars = len(vars_to_plot)
    nrows = math.ceil(n_vars / ncols)
    _, axes = plt.subplots(nrows=nrows, ncols=ncols, figsize=(18, 5 * nrows), dpi=150)
    axes = axes.flatten()

    for i, var in enumerate(vars_to_plot):
        ax = axes[i]
        if var not in df.columns:
            ax.set_title(f"{var} (missing)", fontsize=16)
            ax.axis("off")
            continue

        s = df[var].dropna()
        if pd.api.types.is_numeric_dtype(s) and s.nunique() > 20:
            counts = pd.cut(s, bins=20).value_counts(sort=False)
            x = counts.index.astype(str)
        else:
            counts = s.astype(str).value_counts()
            if len(counts) > 30:
                counts = counts.head(30)
            x = counts.index

        n_bars = len(counts)
        ax.bar(x, counts.values, color=sns.color_palette("tab20", n_colors=n_bars))
        ax.set_title(var, fontsize=16)
        ax.set_ylabel("Count", fontsize=14)
        ax.tick_params(axis="x", rotation=45, labelsize=10)
        ax.tick_params(axis="y", labelsize=12)
        ax.yaxis.set_major_locator(plt.MaxNLocator(nbins=5))
        # ensure top tick ≥ tallest bar
        # top = int(np.nanmax(counts.values)) if len(counts) else 0
        # yloc = plt.MaxNLocator(nbins=5, integer=True)
        # ticks = yloc.tick_values(0, top)
        # ax.set_ylim(0, ticks[-1])
        # ax.yaxis.set_major_locator(yloc)

    for ax in axes[n_vars:]:
        ax.axis("off")

    plt.tight_layout()
    plt.subplots_adjust(bottom=0.06)
    plt.show()


def plot_multi_figures(df: pd.DataFrame):
    # Calls (replace your three plotting cells)
    hist_by(
        df,
        x="age",
        hue="sex",
        title="Age Distribution by Sex",
        xlabel="Age (years)",
    )
    hist_by(
        df,
        x="age",
        hue="site",
        title="Age Distribution by Site",
        xlabel="Age (years)",
    )
    hist_by(
        df,
        x="fd_mean",
        hue="site",
        title="FD Distribution by Site",
        xlabel="Framewise Displacement (mm)",
    )

    plot_var_grid(
        df,
        vars_to_plot=[
            "sex",
            "scanner",
            "tr",
            "te",
            "mb_factor",
            "pa_factor",
            "phase_enc_dir_lr_rl",
            "phase_enc_dir_ap",
            "phase_enc_dir_pa",
            "fixed_effect",
            "volumes",
            "halfpipe",
            "task_length_s",
            "task",
            "target_stimuli",
            "gain",
            "loss",
            "n_trials",
            "anticipation_delay_ms",
            "adaptive_response",
        ],
        ncols=3,
    )
```
