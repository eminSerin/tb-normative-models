```python
import os.path as op

import pandas as pd
```

```python
BASE_DIR = "" # Write your path here


def load_reportvals(path: str) -> pd.DataFrame:
    """
    Load QC reports (i.e., reportvals-*.txt) as a DataFrame by parsing fixed-width columns.

    Parameters
    ----------
    path : str
        Absolute path to `reportvals-*.txt`.

    Returns
    -------
    pd.DataFrame
        Parsed table with headers; the second line of dashes is removed.
        Blank cells are loaded as NaN.
    """
    df = pd.read_fwf(
        path,
        colspecs="infer",
        header=0,
        skiprows=[1],
        na_values=["", "NA", "N/A"],
        keep_default_na=True,
    )
    df.columns = df.columns.str.strip()
    return df


def process_dataset(
    dset_name: str,
    qc_file: str,
    reportvals_path: str,
    reportvals_type: str,
    merge_keys: list[str],
    task_replacements: dict[str, str],
    output_file: str,
    reportvals_rename: dict[str, str] = None,
    reportvals_cols: list[str] = None,
) -> pd.DataFrame:
    """Process a single dataset by merging QC and reportvals data."""
    qc = pd.read_csv(op.join(BASE_DIR, f"dsets/{qc_file}"))

    reportvals_path_full = op.join(BASE_DIR, reportvals_path)
    if reportvals_type == "txt":
        reportvals = load_reportvals(reportvals_path_full).rename(
            columns={"sub": "subject"}
        )
    else:
        reportvals = pd.read_csv(reportvals_path_full)

    if reportvals_rename:
        reportvals = reportvals.rename(columns=reportvals_rename)

    if reportvals_cols is None:
        reportvals_cols = ["subject", "task", "fd_mean"]
    reportvals = reportvals[reportvals_cols]

    qc_comb = qc.merge(reportvals, on=merge_keys, how="left").replace(task_replacements)
    qc_comb.to_csv(op.join(BASE_DIR, f"dsets/{output_file}"), index=False)
    return qc_comb
```

```python
# PIOP1
process_dataset(
    dset_name="piop1",
    qc_file="piop1_qc.csv",
    reportvals_path="qc-reports/exclude-piop1/reportvals-piop1.txt",
    reportvals_type="txt",
    merge_keys=["subject", "task"],
    task_replacements={"workingmemory": "wm"},
    output_file="piop1_qc_fd.csv",
)
```
